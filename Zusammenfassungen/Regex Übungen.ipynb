# Regex-Übungen

Bearbeite die folgenden Aufgaben mit dem Python-Modul `re`. Ergänze jeweils den regulären Ausdruck und teste das Ergebnis.

```python
import re
```

## Aufgabe 1: Telefonnummer

Extrahiere die Telefonnummer aus dem Text.

Erwartetes Ergebnis: `+41 79 123 45 67`

```python
text = "Kontakt: +41 79 123 45 67"

# Ergänze den Regex:
match = re.findall(r"\+[\d ]+", text)
print(match[0])
```

## Aufgabe 2: E-Mail-Adresse

Extrahiere die E-Mail-Adresse aus dem Text.

Erwartetes Ergebnis: `max.mustermann@example.com`

```python
text = "Schreibe an max.mustermann@example.com"

# Ergänze den Regex:
match = re.findall(r"[a-z]*\.[a-z]*@[a-z]*\.com", text)
print(match[0])
```

## Aufgabe 3: Preise

Extrahiere alle Preise ohne die Währung `CHF`.

Erwartetes Ergebnis: `['12.50', '99.90', '5.00']`

```python
text = "Produkte kosten CHF 12.50, CHF 99.90 und CHF 5.00."

# Ergänze den Regex:
match = re.findall(r"(?<=CHF )([\d\.]+)", text)
print(match[0])
```

## Aufgabe 4: Zimmeranzahl

Extrahiere die Zahl vor dem Wort `Zimmer`. Verwende eine positive Lookahead-Bedingung.

Erwartetes Ergebnis: `['3.5']`

```python
text = "Die Wohnung hat 3.5 Zimmer."

# Ergänze den Regex:
match = re.findall(r"([\d\.]+) Zimmer", text)
print(match[0])
```

## Aufgabe 5: Postleitzahl

Extrahiere die Schweizer Postleitzahl.

Erwartetes Ergebnis: `['8001']`

```python
text = "Adresse: Bahnhofstrasse 10, 8001 Zürich"

# Ergänze den Regex:
match = re.findall(r"[0-9]{4}", text)
print(match[0])
```

## Aufgabe 6: Datum

Extrahiere das Datum im Format `TT.MM.JJJJ`.

Erwartetes Ergebnis: `['08.11.2026']`

```python
text = "Die Prüfung findet am 08.11.2026 statt."

# Ergänze den Regex:
match = re.findall(r"[0-9]{2}\.[0-9]{2}\.[0-9]{4}", text)
print(match[0])
```

## Aufgabe 7: Wörter mit Großbuchstaben

Finde alle Wörter, die mit einem Großbuchstaben beginnen.

Erwartetes Ergebnis: `['Python', 'Programmiersprache']`

```python
text = "Python ist eine beliebte Programmiersprache."

# Ergänze den Regex:
match = re.findall(r"[A-Z]{1}[a-z]*", text)
print(match)
```

## Aufgabe 8: Capturing Group

Extrahiere nur die Zahl vor `kW`. Verwende eine Capturing Group und `re.findall()`.

Erwartetes Ergebnis: `['22']`

```python
text = "Die Ladestation hat eine Leistung von 22 kW."

# Ergänze den Regex mit einer Capturing Group:
match = re.findall(r"([\d]+) kW", text)
print(match[0])
```

## Aufgabe 9: Non-Capturing Group

Finde sowohl `http` als auch `https`. Verwende eine Non-Capturing Group.

Hinweis: Die Gruppe soll die beiden Alternativen zusammenfassen, aber nicht als eigene Gruppe gespeichert werden.

Erwartetes Ergebnis: `['http', 'https']`

```python
text = "Webseiten: http://example.com und https://example.org"

# Ergänze den Regex mit (?:...):
match = re.findall(r"(http[s]?)://", text)
print(match)
```

## Aufgabe 10: Fehlende Daten

Extrahiere die Zimmeranzahl aus jeder Zeichenkette. Wenn keine Zimmeranzahl vorhanden ist, soll `None` gespeichert werden. Ersetze anschließend ein deutsches Dezimalkomma durch einen Punkt.

Erwartetes Ergebnis: `['3.5', '4', None, '2.5']`

```python
texts = [
    "3.5 Zimmer",
    "4 Zimmer",
    "Keine Angabe",
    "2,5 Zimmer"
]

rooms = []

for text in texts:
    # Ergänze den Regex und die Behandlung fehlender Treffer:
    match = re.findall(r"([\d,\.]+) Zimmer", text)
    if match:
        rooms.append(match[0])
    else:
        rooms.append(None)

print(rooms)
```
