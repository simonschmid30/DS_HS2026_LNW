# Musterlösung – Arbeitspaket 2

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
raw = pd.read_csv('../../Probepruefung_AP01-03/AP02/products_raw.csv')
clean = raw.copy()
clean['price_chf'] = pd.to_numeric(clean['price_chf'].astype(str).str.replace(',', '.', regex=False), errors='coerce')
clean['rating'] = pd.to_numeric(clean['rating'].astype(str).str.replace(',', '.', regex=False), errors='coerce')
```

### A1–A3
Imputation erhält Stichprobengrösse, kann aber Verteilungen verzerren; Entfernen kann bei vielen oder nicht zufällig fehlenden Werten Bias erzeugen. Korrelation beschreibt einen statistischen Zusammenhang, nicht zwingend eine Ursache. Qualitätsprüfungen sind z.B. Pflichtfelder/Datentypen, Wertebereiche, Duplikate, Schlüsselreferenzen und Plausibilitätschecks.

```python
summary = clean.groupby('category').agg(product_count=('product_id','count'), 
mean_price=('price_chf','mean'), 
median_rating=('rating','median')).sort_values('mean_price', ascending=False)
summary
```

```python
avg = clean.groupby('category')['rating'].mean().dropna()
ax = avg.plot.bar(figsize=(6,4), ylabel='Durchschnittliche Bewertung', title='Bewertung nach Kategorie')
plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```
