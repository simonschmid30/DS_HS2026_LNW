# Musterlösung – Arbeitspaket 3

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.metrics import mean_absolute_error, accuracy_score
df = pd.read_csv('../../Probepruefung_AP01-03/AP03/bikes.csv')
features = ['age','km','battery_wh','condition']
X = df[features]
X_train, X_test, y_train, y_test = train_test_split(X, df['price_chf'], test_size=.2, random_state=42)
reg = DecisionTreeRegressor(max_depth=4, random_state=42)
reg.fit(X_train, y_train)
print(mean_absolute_error(y_train, reg.predict(X_train)), mean_absolute_error(y_test, reg.predict(X_test)))
```

### A1–A3
Overfitting bedeutet, dass ein Modell Trainingsrauschen lernt und auf neuen Daten schlecht generalisiert. Gegenmassnahmen sind z.B. geringere Baumtiefe, Regularisierung, Cross-Validation und mehr Daten. Der Testdatensatz bleibt unberührt, damit seine Metrik eine faire Schätzung der Generalisierung liefert. MAE misst Fehler einer kontinuierlichen Zielgrösse wie Preis; Accuracy misst Anteil korrekter Klassen wie `electric`.

```python
Xc = df[features]
yc = df['electric']
Xtr, Xte, ytr, yte = train_test_split(Xc, yc, test_size=.2, random_state=42, stratify=yc)
clf = DecisionTreeClassifier(max_depth=3, random_state=42)
clf.fit(Xtr, ytr)
print(accuracy_score(yte, clf.predict(Xte)))
```

```python
for depth in [2,4,None]:
    m = DecisionTreeRegressor(max_depth=depth, random_state=42).fit(X_train, y_train)
    print(depth, mean_absolute_error(y_train,m.predict(X_train)), mean_absolute_error(y_test,m.predict(X_test)))
# Tiefe None overfitten typischerweise stärker; die Test-MAE ist entscheidend.
```
