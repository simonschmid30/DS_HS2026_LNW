### A1–A3
Primärschlüssel identifizieren einen Datensatz eindeutig (z.B. `events.event_id`); Fremdschlüssel verweisen auf einen Primärschlüssel einer anderen Tabelle (z.B. `events.venue_id`). Die Beziehung ist 1:n, weil ein Veranstaltungsort mehrere Events hosten kann, ein Event aber genau einen Veranstaltungsort hat. `WHERE` filtert Zeilen vor der Gruppierung, `HAVING` filtert Gruppen nach Aggregation.

```python
import sqlite3
con = sqlite3.connect("../../Probepruefung_AP01-03/AP01/festival.db")
```

### B1

```python
print(con.execute("""
SELECT e.title, COALESCE(SUM(b.tickets), 0) AS sold
FROM events e
LEFT JOIN bookings b ON e.event_id = b.event_id
GROUP BY e.event_id, e.title
HAVING COALESCE(SUM(b.tickets), 0) > 2;
""").fetchall())
```

### B2

```python
print(con.execute("""
SELECT v.city, SUM(b.tickets * b.price) AS revenue
FROM venues v 
JOIN events e ON v.venue_id=e.venue_id
JOIN bookings b ON e.event_id=b.event_id
WHERE e.event_date >= '2026-03-13'
GROUP BY v.city 
ORDER BY revenue DESC;
""").fetchall())
```

### C1

```python
print(con.execute("""
SELECT e.title, e.genre, v.name AS venue
FROM events e 
JOIN venues v ON e.venue_id=v.venue_id
WHERE v.city='Bern' 
ORDER BY e.title;
""").fetchall())
```

### C2

```python
print(con.execute("""
SELECT 
    v.membership, 
    SUM(b.tickets) AS tickets, 
    SUM(b.tickets*b.price) AS revenue
FROM visitors v 
JOIN bookings b ON v.visitor_id=b.visitor_id
GROUP BY v.membership 
HAVING SUM(b.tickets)>=3;
""").fetchall())
```

### C3

```python
print(con.execute("""
SELECT v.name, SUM(b.tickets) AS tickets, SUM(b.tickets*b.price) AS revenue
FROM visitors v 
JOIN bookings b ON v.visitor_id=b.visitor_id
GROUP BY v.visitor_id, v.name 
ORDER BY revenue DESC LIMIT 3;
""").fetchall())
```
